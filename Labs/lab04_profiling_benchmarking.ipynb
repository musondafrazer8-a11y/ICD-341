### Task 4 Comparison

slow_function (loop):
- Uses Python for-loop + list append + Python sum()
- 100k elements = 20.4 ms → 1M would be ~200ms
- Profiling shows loop body (append) takes >85% time

fast_function (vectorized):
- Uses NumPy x**2 and np.sum in compiled C
- 100k = 85 µs, 1M = ~1 ms
- ~200-300x faster

Conclusion: Vectorization avoids Python interpreter overhead. Always use NumPy for large data.

```python
import numpy as np
np.random.seed(42)

data = np.random.normal(
    loc=100,
    scale=20,
    size=5_000_000
)
print("Data generated:", data.shape)
```

```python
def squared_deviations_loop(x):
    mean = np.mean(x)
    total = 0
    for value in x:
        total += (value - mean) ** 2
    return total

def squared_deviations_numpy(x):
    mean = np.mean(x)
    return np.sum((x - mean) ** 2)
```

```python
%timeit squared_deviations_numpy(data)
```

```python
x_small = data[:100_000]  # 100k sample for loop

%timeit squared_deviations_loop(x_small)
```

```python
%timeit squared_deviations_numpy(x_small)
```

```python
# Verify both give same answer
print("Loop result:", squared_deviations_loop(x_small))
print("NumPy result:", squared_deviations_numpy(x_small))

# Speedup calculation
# 46.9 ms / 0.8 ms ≈ 58x faster
```

1. What is vectorization?
Vectorization is the process of performing operations on entire arrays/vectors at once instead of element-by-element using Python loops. NumPy applies the operation in compiled C code internally.

2. Why is vectorization useful?
.a) Speed:* 10x to 100x faster than Python loops - uses optimized C and SIMD.
.b) Cleaner code:* 1 line `y = 2*x + 5` instead of 4-line for-loop.
.c) Memory efficiency:* Avoids creating intermediate Python objects.
Essential for statistical computing with large data.

3. Difference between Python loop and NumPy vectorized operation?*
Python Loop | NumPy Vectorized
Runs in Python interpreter, slow | Runs in compiled C, fast
Explicit `for i in range(len(x)):` | Implicit, no Python-level loop: `2*x+5`
Type checking per element | Operates on homogeneous `dtype` block
More code, error-prone | Short, readable, less bug

3. Why might NumPy be faster?

.a) Compiled C code:* NumPy operations are written in C/Fortran, not Python. No Python type-checking per element.
.b) No Python loops:* Loops run in interpreter overhead; NumPy uses vectorized SIMD CPU instructions.
.c) Memory locality:* NumPy arrays are contiguous blocks of same `dtype` in RAM, fits CPU cache. Python lists are pointers scattered.
.d) Broadcasting avoids copies:* It virtually stretches `1x3` to `3x3` without actually allocating new memory.
> NumPy is faster because it pushes the heavy work to optimized low-level code and avoids the overhead of the Python interpreter loop, which is critical for large statistical datasets.

4. Explain why vectorization can produce such a large performance difference.

- Python loops run in interpreter with type checking, function calls and object overhead for each element.
- NumPy is compiled C code operating on contiguous memory blocks of fixed dtype.
- Uses SIMD and optimized BLAS libraries, CPU cache locality, and avoids creating Python objects per element.
- Broadcasting does not copy data, it uses stride tricks.

That is why difference is 100x-1000x for large statistical data.

5. How could profiling help you decide where to optimize a statistical program?

> Profiling measures where the program spends most time (hotspots). Instead of guessing and optimizing everything, you use `cProfile`, `line_profiler` or `%timeit` to find the bottleneck - e.g., 90% of time in one loop. Then you optimize only that part with vectorization/Numba. This saves development time and keeps code readable where performance doesn't matter.
