```python
import numpy as np

# ==========================================
# PART A: Basic vectorization
# ==========================================
print("--- PART A ---")
x = np.array([1, 2, 3, 4, 5])
y = 2 * x + 5
print("y = 2x + 5:", y)


# ==========================================
# PART B: Vectorized statistical transformation
# ==========================================
print("\n--- PART B ---")
marks = np.array([40, 55, 62, 70, 81, 90])
adjusted = marks + 5
percentage = marks / 100

# Standardization (z-score)
z = (marks - np.mean(marks)) / np.std(marks)

print("Adjusted marks:", adjusted)
print("Percentage marks:", percentage)
print("Standardized marks (z):", z)
print("Mean of z (should be ~0):", z.mean())


# ==========================================
# PART C & D: Broadcasting
# ==========================================
print("\n--- PART C & D ---")
data = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [90, 85, 95]
])
adjustment = np.array([5, 0, 2])
result = data + adjustment
print("Broadcasted addition:\n", result)

# Centering the data
data_stats = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])
means = data_stats.mean(axis=0)
centered = data_stats - means
print("\nCentered data:\n", centered)
print("Verify centered means (should be ~[0, 0, 0]):", centered.mean(axis=0))


# ==========================================
# PART E: Vectorization vs Loops
# ==========================================
print("\n--- PART E ---")
x_large = np.arange(1_000_000)

# Using a loop
result_loop = []
for value in x_large:
    result_loop.append(value**2 + 2*value + 1)

# Using vectorization
result_vectorized = x_large**2 + 2*x_large + 1

print("Do they match?", np.allclose(result_loop, result_vectorized))


# ==========================================
# PART F: Statistical computing application
# ==========================================
print("\n--- PART F ---")
np.random.seed(123)
x_norm = np.random.normal(100, 15, 100_000)
z_norm = (x_norm - x_norm.mean()) / x_norm.std()

print("Mean of standardized data (should be ~0):", z_norm.mean())
print("Std of standardized data (should be ~1):", z_norm.std())
```

```python
# ==========================================
# LAB 3 CHALLENGE
# ==========================================
print("\n--- LAB 3 CHALLENGE ---")

def standardize(arr):
    return (arr - np.mean(arr)) / np.std(arr)

# 1. Test 1D arrays
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([5, 5, 5, 5, 5]) 
arr3 = np.array([1.5, 2.5, 3.5, 4.5])

print("Test 1:", standardize(arr1))
print("Test 3:", standardize(arr3))
# Note: We skip arr2 because the standard deviation is 0, which causes a division-by-zero error.

# 2. Modify to standardize a 2D dataset column-by-column
def standardize_columns(matrix):
    return (matrix - np.mean(matrix, axis=0)) / np.std(matrix, axis=0)

matrix = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000]
])

standardized_matrix = standardize_columns(matrix)
print("\nStandardized Matrix (column-wise):\n", standardized_matrix)

# Verify the result
print("\nMean of columns (should be [0, 0, 0]):", np.mean(standardized_matrix, axis=0))
print("Std of columns (should be [1, 1, 1]):", np.std(standardized_matrix, axis=0))
```

Part A (Vectorization):

· What is vectorization? Vectorization is the process of executing operations on entire arrays at once, rather than using a Python for loop to iterate through elements one by one.
· Why is it useful? It makes code shorter, easier to read, and executes significantly faster because the underlying operations are written in optimized C code rather than interpreted Python.
· Difference between loop and vectorized? A Python loop iterates element-by-element with high overhead. A vectorized operation hands the whole array to NumPy's backend, processing data in chunks and utilizing CPU-level optimizations (SIMD).

Part C (Broadcasting):

· Why does data + adjustment work? This works due to broadcasting. NumPy automatically stretches the smaller 1x3 array (adjustment) across the rows of the 3x3 array (data) so they have compatible shapes for element-wise addition.

Part E (Loops vs Vectorization):

· Are the results equal? Yes, np.allclose() returns True.
· Which is easier to read? The vectorized version (x**2 + 2*x + 1), as it looks exactly like the mathematical formula.
· Which is faster? The vectorized version.
· Why? It avoids Python's per-iteration interpreter overhead, utilizing compiled C code and contiguous memory blocks instead.
