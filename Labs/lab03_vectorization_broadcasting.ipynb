```python
import numpy as np

marks = np.array([40, 55, 62, 70, 81, 90])

# lecturer adds 5 marks
adjusted = marks + 5
percentage = marks / 100

print("marks:", marks)
print("adjusted +5:", adjusted)
print("percentage:", percentage)

# Task 2 - standardization  z = (x - mean) / std
z = (marks - np.mean(marks)) / np.std(marks)

print("\nz-scores:", z)
print("mean of z:", np.mean(z))  # should be ~0
print("std of z:", np.std(z))    # should be 1
```

```python
import numpy as np

data = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

means = data.mean(axis=0)
centered = data - means

print("data:\n", data)
print("\ncolumn means:", means)  # [20. 30. 40.]
print("\ncentered = data - means:\n", centered)

# Task 3 verify
print("\nVerify mean of centered columns:", centered.mean(axis=0))
```

```python
import numpy as np
import time

x = np.arange(1_000_000)

# 1. Loop version
start = time.time()
result_loop = []
for value in x:
    result_loop.append(value**2 + 2*value + 1)
loop_time = time.time() - start
print(f"Loop time: {loop_time:.2f} seconds")

# 2. Vectorized version
start = time.time()
result_vectorized = x**2 + 2*x + 1
vec_time = time.time() - start
print(f"Vectorized time: {vec_time:.4f} seconds")

# 3. Check they agree
print("Same result?", np.allclose(result_loop, result_vectorized))
print(f"Speedup: {loop_time / vec_time:.1f}x faster")
```

```python
import numpy as np

def standardize(x):
    return (x - np.mean(x)) / np.std(x)

# Test 1 - small marks
arr1 = np.array([40, 55, 62, 70, 81, 90])
print("arr1:", standardize(arr1))
print("mean:", standardize(arr1).mean(), "std:", standardize(arr1).std())

# Test 2 - random normal 100, 15
np.random.seed(0)
arr2 = np.random.normal(100, 15, 1000)
print("\narr2 sample:", standardize(arr2)[:5])
print("mean:", standardize(arr2).mean(), "std:", standardize(arr2).std())

# Test 3 - 2D? Use axis? For now 1D
arr3 = np.array([10, 20, 30, 40, 50])
print("\narr3:", standardize(arr3))
print("mean:", standardize(arr3).mean(), "std:", standardize(arr3).std())
```
