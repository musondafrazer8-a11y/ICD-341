```python
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])
np.mean(marks)
np.median(marks)
np.std(marks)
np.min(marks)
np.max(marks)
np.sum(marks)

print("mean:", np.mean(marks))
print("median:", np.median(marks))
print("std:", np.std(marks))
print("min:", np.min(marks))
print("max:", np.max(marks))
print("sum:", np.sum(marks))
```

```python
data = np.array([
 [45, 67, 82],
 [56, 91, 73],
 [64, 88, 76]
])
print(data.shape)
print(data.ndim)
print("\n1. First row:", data[0])
print("2. Second column:", data[:, 1])
print("3. Value 91:", data[1, 1])
print("4. First two rows:\n", data[0:2])
print("5. Last two columns:\n", data[:, 1:])
```

```python
np.mean(data, axis=0)
# data is same 3x3
# Rows = Students, Columns = Maths | Stats | Programming

mean_per_subject = np.mean(data, axis=0)
mean_per_student = np.mean(data, axis=1)

print("Mean per subject (Maths, Stats, Programming):", mean_per_subject)
print("Mean per student (Student1,2,3):", mean_per_student)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# set seed so results are reproducible for marking
np.random.seed(0)
x = np.random.normal(loc=50, scale=10, size=1000)

print(f"mean: {np.mean(x):.2f}")
print(f"median: {np.median(x):.2f}")
print(f"std: {np.std(x):.2f}")
print(f"min: {np.min(x):.2f}")
print(f"max: {np.max(x):.2f}")

plt.hist(x, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```
